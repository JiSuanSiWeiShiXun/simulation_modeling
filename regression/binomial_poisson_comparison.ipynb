# Binomial and Poisson Distribution Comparison

**Objective:** Generate 1,000 random samples from a Binomial distribution with parameters n=300 and p=0.01, and from a Poisson distribution with parameter λ=np. Compare their shapes and evaluate whether the Poisson distribution provides a good approximation to the Binomial distribution.

## Problem Statement
- Binomial Distribution: B(n=300, p=0.01)
- Poisson Distribution: Poisson(λ = np = 3)
- Sample Size: 1,000 samples

## Theoretical Background

**Poisson Approximation to Binomial:**
When n is large and p is small, the Binomial(n, p) distribution can be approximated by Poisson(λ=np).

**Conditions:**
1. n ≥ 20 (large number of trials)
2. p ≤ 0.05 (small probability)
3. np < 10 (moderate expected value)

## 1. Import Required Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
import pandas as pd

# Set random seed for reproducibility
np.random.seed(2025)

print("Libraries imported successfully!")
```

## 2. Define Parameters and Generate Samples

```python
# Parameters
n = 300          # Number of trials
p = 0.01         # Probability of success
lambda_param = n * p  # Poisson parameter λ = np = 3
sample_size = 1000    # Number of samples

print("="*60)
print("Binomial vs Poisson Distribution Comparison")
print("="*60)
print(f"Parameters:")
print(f"  Binomial: n = {n}, p = {p}")
print(f"  Poisson:  λ = np = {lambda_param}")
print(f"  Sample size: {sample_size}")
```

```python
# Generate random samples
binomial_samples = np.random.binomial(n, p, sample_size)
poisson_samples = np.random.poisson(lambda_param, sample_size)

print(f"\nGenerated {sample_size} samples from each distribution")
print(f"Binomial samples range: [{binomial_samples.min()}, {binomial_samples.max()}]")
print(f"Poisson samples range: [{poisson_samples.min()}, {poisson_samples.max()}]")
```

## 3. Calculate Descriptive Statistics

```python
# Calculate statistics
stats_df = pd.DataFrame({
    'Statistic': ['Mean', 'Variance', 'Std Deviation', 'Min', 'Max', 'Median'],
    'Binomial': [
        np.mean(binomial_samples),
        np.var(binomial_samples, ddof=1),
        np.std(binomial_samples, ddof=1),
        np.min(binomial_samples),
        np.max(binomial_samples),
        np.median(binomial_samples)
    ],
    'Poisson': [
        np.mean(poisson_samples),
        np.var(poisson_samples, ddof=1),
        np.std(poisson_samples, ddof=1),
        np.min(poisson_samples),
        np.max(poisson_samples),
        np.median(poisson_samples)
    ]
})

stats_df['Difference'] = abs(stats_df['Binomial'] - stats_df['Poisson'])

print("\nDescriptive Statistics:")
print(stats_df.to_string(index=False))
```

```python
# Theoretical values
theoretical_mean = n * p
theoretical_var_binom = n * p * (1 - p)
theoretical_var_poisson = lambda_param

theoretical_df = pd.DataFrame({
    'Parameter': ['Mean', 'Variance (Binomial)', 'Variance (Poisson)'],
    'Theoretical Value': [theoretical_mean, theoretical_var_binom, theoretical_var_poisson]
})

print("\nTheoretical Values:")
print(theoretical_df.to_string(index=False))
```

## 4. Visualize Distributions (Side-by-Side Subplots)

```python
# Create side-by-side subplots
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Subplot 1: Binomial Distribution
axes[0].hist(binomial_samples, bins=range(0, max(binomial_samples)+2), 
             density=True, alpha=0.7, color='steelblue', edgecolor='black')
axes[0].set_title(f'Binomial Distribution\n(n={n}, p={p})', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Value', fontsize=11)
axes[0].set_ylabel('Probability Density', fontsize=11)
axes[0].grid(axis='y', alpha=0.3)
axes[0].axvline(np.mean(binomial_samples), color='red', linestyle='--', 
                linewidth=2, label=f'Mean = {np.mean(binomial_samples):.2f}')
axes[0].legend()

# Subplot 2: Poisson Distribution
axes[1].hist(poisson_samples, bins=range(0, max(poisson_samples)+2), 
             density=True, alpha=0.7, color='coral', edgecolor='black')
axes[1].set_title(f'Poisson Distribution\n(λ={lambda_param})', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Value', fontsize=11)
axes[1].set_ylabel('Probability Density', fontsize=11)
axes[1].grid(axis='y', alpha=0.3)
axes[1].axvline(np.mean(poisson_samples), color='red', linestyle='--', 
                linewidth=2, label=f'Mean = {np.mean(poisson_samples):.2f}')
axes[1].legend()

plt.tight_layout()
plt.show()
```

## 5. Overlay Comparison Plot

```python
# Overlay both distributions
fig, ax = plt.subplots(figsize=(10, 6))

bins = range(0, max(max(binomial_samples), max(poisson_samples))+2)
ax.hist(binomial_samples, bins=bins, density=True, alpha=0.5, 
        color='steelblue', label=f'Binomial(n={n}, p={p})', edgecolor='black')
ax.hist(poisson_samples, bins=bins, density=True, alpha=0.5, 
        color='coral', label=f'Poisson(λ={lambda_param})', edgecolor='black')

ax.set_title('Binomial vs Poisson Distribution Overlay', fontsize=14, fontweight='bold')
ax.set_xlabel('Value', fontsize=12)
ax.set_ylabel('Probability Density', fontsize=12)
ax.legend(fontsize=11)
ax.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

## 6. Statistical Hypothesis Test

```python
# Kolmogorov-Smirnov test
ks_statistic, ks_pvalue = stats.ks_2samp(binomial_samples, poisson_samples)

print("\nKolmogorov-Smirnov Test:")
print("="*60)
print(f"KS Statistic: {ks_statistic:.4f}")
print(f"P-value: {ks_pvalue:.4f}")
print()

if ks_pvalue > 0.05:
    print("✓ Result: Distributions are NOT significantly different (p > 0.05)")
    print("  → Poisson provides a GOOD approximation")
else:
    print("⚠ Result: Distributions are significantly different (p ≤ 0.05)")
    print("  → Poisson approximation may be limited")
```

## 7. Evaluate Approximation Quality

```python
# Calculate relative errors
mean_relative_error = abs(np.mean(binomial_samples) - np.mean(poisson_samples)) / theoretical_mean * 100
var_relative_error = abs(np.var(binomial_samples, ddof=1) - theoretical_var_poisson) / theoretical_var_poisson * 100

print("\nApproximation Quality Assessment:")
print("="*60)
print(f"\nConditions for Poisson Approximation:")
print(f"  1. n should be large (n ≥ 20): n = {n} ✓")
print(f"  2. p should be small (p ≤ 0.05): p = {p} ✓")
print(f"  3. np should be moderate (< 10): np = {lambda_param} ✓")
print(f"\nRelative Errors:")
print(f"  - Mean: {mean_relative_error:.2f}%")
print(f"  - Variance: {var_relative_error:.2f}%")
print()

if mean_relative_error < 5 and var_relative_error < 10:
    print("✓ CONCLUSION: The Poisson distribution provides an EXCELLENT")
    print("  approximation to the Binomial distribution under these conditions.")
else:
    print("⚠ CONCLUSION: The Poisson approximation is acceptable but shows")
    print("  some deviation from the Binomial distribution.")
```

## 8. Probability Mass Function Comparison

```python
# Compare theoretical PMFs
x_values = np.arange(0, 15)
binom_pmf = stats.binom.pmf(x_values, n, p)
poisson_pmf = stats.poisson.pmf(x_values, lambda_param)

plt.figure(figsize=(10, 6))
plt.plot(x_values, binom_pmf, 'o-', color='steelblue', linewidth=2, 
         markersize=8, label=f'Binomial(n={n}, p={p})')
plt.plot(x_values, poisson_pmf, 's--', color='coral', linewidth=2, 
         markersize=8, label=f'Poisson(λ={lambda_param})')

plt.title('Theoretical PMF Comparison', fontsize=14, fontweight='bold')
plt.xlabel('Value', fontsize=12)
plt.ylabel('Probability', fontsize=12)
plt.legend(fontsize=11)
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

# Show PMF values in a table
pmf_comparison = pd.DataFrame({
    'Value': x_values[:10],
    'Binomial PMF': binom_pmf[:10],
    'Poisson PMF': poisson_pmf[:10],
    'Difference': abs(binom_pmf[:10] - poisson_pmf[:10])
})

print("\nPMF Comparison (First 10 Values):")
print(pmf_comparison.to_string(index=False))
```

## 9. Summary and Discussion

### Key Findings:

1. **Shape Comparison:**
   - Both distributions exhibit similar bell-shaped patterns
   - Centers around λ = np = 3
   - Both are slightly right-skewed

2. **Statistical Measures:**
   - Means are very close
   - Variances are comparable
   - KS test confirms similarity

3. **Approximation Quality:**
   - All conditions satisfied (n large, p small, np moderate)
   - Relative errors are minimal
   - Poisson provides excellent approximation

### Conclusion:

**YES, the Poisson distribution provides a GOOD approximation** to the Binomial distribution under these conditions:
- n = 300 (large)
- p = 0.01 (small)
- λ = np = 3 (moderate)

The Poisson approximation is particularly useful because:
- Simpler to compute (only 1 parameter vs 2)
- Easier to work with analytically
- Provides nearly identical results for practical purposes
